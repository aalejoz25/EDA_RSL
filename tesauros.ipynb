# Tesauros y su Uso en la Investigación

Los tesauros son vocabularios controlados y estructurados que ayudan a organizar y estandarizar términos en un campo específico. Son fundamentales para realizar búsquedas sistemáticas de literatura.

## Tesauros estandarizados:

### 1. AGROVOC (FAO)
- **Descripción**: Tesauro multilingüe de la FAO que cubre agricultura, silvicultura, pesca y alimentación
- **Utilidad**: Excelente para términos relacionados con cultivos y agricultura
- **Acceso**: [AGROVOC](http://aims.fao.org/vest-registry/vocabularies/agrovoc)

### 2. CAB Thesaurus
- **Descripción**: Especializado en ciencias de la vida, incluyendo agricultura y ciencias vegetales
- **Utilidad**: Muy útil para términos sobre enfermedades de plantas y cultivos
- **Acceso**: [CAB Thesaurus](https://www.cabi.org/cabthesaurus/)

### 3. IEEE Thesaurus
- **Descripción**: Vocabulario controlado para términos técnicos y de ingeniería
- **Utilidad**: Perfecto para términos de IA y procesamiento de datos
- **Acceso**: Disponible a través de IEEE Xplore
- [IEEE Thesaurus](https://ieee-org.widen.net/s/ljqzpfxm5g/ieee-thesaurus)

### 4. NAL Agricultural Thesaurus
- **Descripción**: Tesauro de la Biblioteca Nacional de Agricultura de EE.UU.
- **Utilidad**: Específico para términos agrícolas y de cultivos
- **Acceso**: [NAL Thesaurus](https://agclass.nal.usda.gov/)

### Beneficios de Usar Tesauros en la Búsqueda:
1. **Exhaustividad**: Asegura que no se pierdan documentos relevantes
2. **Precisión**: Reduce el ruido en los resultados
3. **Consistencia**: Mantiene coherencia en la terminología
4. **Jerarquía**: Permite expandir o reducir el alcance de la búsqueda

## Uso de los tesauros

### Pasos para una Búsqueda Efectiva:

1. **Identificar Conceptos Clave o Macro**

2. **Consultar Tesauros Específicos**
   - AGROVOC: Para términos agrícolas
   - CAB Thesaurus: Para enfermedades de plantas
   - IEEE Thesaurus: Para términos de IA y asociados a ingenieria
   
3. **Construir Jerarquías de Términos**
   - Identificar términos más específicos
   - Identificar términos relacionados
   - Identificar términos más amplios

4. **Combinar Términos**
   - Usar operadores booleanos (AND, OR)
   - Incluir variaciones de términos
   - Considerar sinónimos y términos relacionados

# Resumen de Tesauros: ScienceDirect

## Definición y Evolución
Los tesauros han evolucionado desde su concepción original como "tesoro de palabras" hasta convertirse en herramientas sofisticadas para la organización y recuperación de información. Según las definiciones modernas, un tesauro es:

- Un vocabulario controlado y dinámico
- Una herramienta que establece relaciones semánticas y sintácticas entre términos
- Un instrumento para el control terminológico en sistemas de información

## Características Principales
1. **Estructura Formal**
   - Relaciones jerárquicas entre conceptos
   - Relaciones semánticas entre términos
   - Control de vocabulario
   - Organización sistemática

2. **Funcionalidades**
   - Control terminológico
   - Ayuda en la recuperación de información
   - Expansión de consultas
   - Normalización de vocabulario

## Tipos de Tesauros
1. **Por Alcance**
   - Generales
   - Especializados
   - Multidisciplinarios
   - Monodisciplinarios

2. **Por Estructura**
   - Alfabéticos
   - Sistemáticos (temáticos)
   - Gráficos

3. **Por Tamaño**
   - Macrotesauros
   - Microtesauros

4. **Por Idioma**
   - Monolingües
   - Bilingües
   - Plurilingües

## Control Terminológico
Los tesauros pueden ser:
1. **Controlados**
   - Un solo término por concepto
   - Relaciones estandarizadas
   - Significados delimitados

2. **De lenguaje libre**
   - Múltiples términos por concepto
   - Mayor flexibilidad
   - Basados en literatura relevante

## Usos Modernos
1. Indexación de documentos
2. Búsqueda y recuperación de información
3. Apoyo en sistemas de IA y procesamiento de lenguaje natural
4. Organización del conocimiento
5. Desarrollo de ontologías y taxonomías

## Facetas Principales

### Faceta 1: TECNOLOGÍA (Herramientas y Métodos)
```

```

### Faceta 2: ENFERMEDADES (Contexto/Problema)
```

 "plant health" OR "crop infection*")
```

### Faceta 3: PREDICCIÓN Y MODELADO (Acción/Propósito)
```
("predict*" OR "forecast*" OR "model*" OR "estimation" OR 
 "early detection" OR "assessment" OR "monitoring" OR
 "analysis" OR "evaluation")
```

### Faceta 4: OBJETO DE ESTUDIO (Objetivo Final)
```
("crop yield" OR "fruit quality" OR "harvest quality" OR 
 "production quality" OR "crop performance")
```

## Justificacion del orden de las facetas

1. **Tecnología** (¿Con qué herramientas?)
   - Define las herramientas y métodos disponibles
   - Base tecnológica para la solución

2. **Enfermedades** (¿Cuál es el problema?)
   - Define el contexto del problema
   - Establece el ámbito de aplicación
   - Identifica la variable independiente principal

3. **Predicción** (¿Qué queremos hacer?)
   - Define la acción sobre el problema
   - Establece el tipo de análisis
   - Conecta el problema con el objetivo

4. **Objeto de Estudio** (¿Qué queremos impactar?)
   - Define el objetivo final
   - Establece las variables dependientes
   - Determina los resultados esperados

## Ejemplo de Construcción Progresiva Refinada

1. **Base Tecnológica**:
   ```
   ("artificial intelligence" OR "machine learning" OR "computer vision")
   ```

2. **+ Contexto del Problema**:
   ```
   [Nivel 1] AND ("plant disease*" OR "crop disease*" OR "pathogen*")
   ```

3. **+ Propósito de Análisis**:
   ```
   [Nivel 2] AND ("predict*" OR "forecast*" OR "model*")
   ```

4. **+ Objetivo Final**:
   ```
   [Nivel 3] AND ("crop yield" OR "fruit quality")
   ```

## ¿Como afecta el orden logico al analisis?

1. **Flujo Lógico**:
   - De herramientas → problema → acción → objetivo
   - Sigue un proceso de pensamiento natural
   - Facilita la comprensión del alcance

2. **Claridad en las relaciones**:
   - Vincula directamente el problema con su solución
   - Establece clara relación causa-efecto
   - Mejora la identificación de variables

3. **Mayor Precisión**:
   - Enfoca primero el contexto del problema
   - Clarifica el tipo de análisis necesario
   - Define mejor el impacto esperado
```

Este orden refleja mejor la secuencia lógica de:
1. Con qué herramientas contamos
2. Qué problema enfrentamos
3. Qué queremos hacer al respecto
4. Qué resultados esperamos obtener

# Construcción de Ecuaciones de Búsqueda usando Tesauros Estandarizados

Este notebook demuestra cómo usar la clase `ThesaurusBuilder` para construir ecuaciones de búsqueda que cumplan con los estándares de tesauros.

## Relaciones Básicas en un Tesauro

Las relaciones fundamentales en un tesauro estandarizado son:

1. **BT (Broader Term / Término Más Amplio)**
   - Indica una relación jerárquica hacia arriba
   - El término BT es más general que el término actual
   - Ejemplo: "machine learning" BT "artificial intelligence"

2. **NT (Narrower Term / Término Más Específico)**
   - Indica una relación jerárquica hacia abajo
   - El término NT es más específico que el término actual
   - Ejemplo: "deep learning" NT "machine learning"

3. **RT (Related Term / Término Relacionado)**
   - Indica una asociación entre términos que no es jerárquica
   - Los términos RT tienen una relación semántica o conceptual
   - Ejemplo: "supervised learning" RT "unsupervised learning"

4. **UF (Used For / Usado Por)**
   - Indica que el término actual es el preferido
   - Los términos UF son sinónimos o variantes no preferidas
   - Ejemplo: "artificial intelligence" UF "AI"

5. **USE (Usar)**
   - Indica que existe un término preferido
   - Dirige desde un término no preferido al término autorizado
   - Ejemplo: "AI" USE "artificial intelligence"

6. **SN (Scope Note / Nota de Alcance)**
   - No es una relación, sino una nota explicativa
   - Define el significado y uso del término en el tesauro
   - Ejemplo: SN "Systems that can learn from data" para "machine learning"

Estas relaciones son bidireccionales:
- Si A es BT de B, entonces B es NT de A
- Si A tiene UF B, entonces B tiene USE A
- Si A es RT de B, entonces B es RT de A

### Fuentes y Estándares

Esta estructura de relaciones está basada en estándares internacionales y documentación especializada:

1. **ISO 25964-1:2011** - "Information and documentation — Thesauri and interoperability with other vocabularies — Part 1: Thesauri for information retrieval"
   - Principal estándar internacional para la construcción de tesauros
   - Define la estructura y relaciones básicas
   - Establece las buenas prácticas para el control de vocabulario

2. **ANSI/NISO Z39.19-2005 (R2010)** - "Guidelines for the Construction, Format, and Management of Monolingual Controlled Vocabularies"
   - Guía americana para la construcción de vocabularios controlados
   - Define las relaciones semánticas y su uso
   - Establece principios para la gestión de tesauros

3. **UNESCO Thesaurus Guidelines** (2005)
   - Guía práctica para la construcción de tesauros multilingües
   - Define la implementación de relaciones jerárquicas y asociativas
   - Proporciona ejemplos de uso en contextos internacionales

4. **ASIS&T Thesaurus Construction Guidelines** - American Society for Information Science and Technology
   - Directrices para la construcción y mantenimiento de tesauros
   - Define las mejores prácticas para establecer relaciones
   - Proporciona ejemplos prácticos de implementación

5. **The Thesaurus: Review, Renaissance, and Revision** (2004)
   - Autores: Sandra K. Roe y Alan R. Thomas
   - Analiza la evolución y modernización de los tesauros
   - Discute la implementación práctica de las relaciones

Estas fuentes establecen los estándares internacionales que aseguran:
- Consistencia en la estructura del tesauro
- Interoperabilidad entre diferentes sistemas
- Calidad en el control del vocabulario
- Correcta gestión de las relaciones semánticas

## Tesauros más relevantes para este proyecto (y cómo usarlos por facetas)

Para **predicción de rendimiento y calidad de frutos en cultivos de ciclo corto afectados por enfermedades mediante IA**, lo más consistente es combinar tesauros por dominio:

1. **IEEE Thesaurus** (tecnología)
   - Para: *Artificial intelligence, Machine learning, Deep learning, Computer vision, Image processing, Remote sensing, Neural networks, Predictive analytics*.
   - Uso: cubrir métodos/algoritmos, visión por computador, sensores y análisis de datos.

2. **AGROVOC (FAO)** (agricultura)
   - Para: *cultivos/horticultura*, *enfermedades de plantas*, *producción/rendimiento*, *calidad* (incluida calidad nutricional).
   - Uso: normalizar el vocabulario agrícola y evitar variaciones locales de términos.

3. **NAL Agricultural Thesaurus (USDA)** (agricultura + patología + calidad)
   - Para: términos finos de *plant diseases*, *disease severity*, *crop yield*, *fruit quality* y atributos (p. ej. sólidos solubles, firmeza).
   - Además: muy útil para *nutritional quality* y variables de matriz nutricional (macro/micro/minerales/vitaminas).

4. **CAB Thesaurus (CABI)** (ciencias de la vida / fitopatología)
   - Para: patógenos, síntomas, tipos de enfermedad, diagnóstico/detección.
   - Uso: muy útil para capturar literatura de fitopatología aplicada.

5. **ACM Computing Classification System (ACM CCS)** (computación, complementario a IEEE)
   - Para: taxonomía de métodos (aprendizaje supervisado, visión por computador, etc.).

> Recomendación práctica: usa estos tesauros como **fuente** para elegir *términos preferidos* y *sinónimos* (UF/USE). En el builder puedes mantener el término preferido y agregar sinónimos como `USE`.

````python
# Tesauro por facetas (proyecto): IA + enfermedades + predicción + NUTRICIÓN (+ rendimiento opcional)
from typing import Optional, Sequence
from thesaurus_builder import ThesaurusBuilder, RelationType

def add_preferred_with_synonyms(
    builder: ThesaurusBuilder,
    preferred: str,
    synonyms: Optional[Sequence[str]] = None,
    scope_note: Optional[str] = None,
):
    term = builder.add_term(preferred, scope_note=scope_note, is_preferred=True)
    for syn in (synonyms or []):
        syn_term = builder.add_term(syn, is_preferred=False)
        syn_term.add_relation(RelationType.USE, term)
    return term


builder = ThesaurusBuilder()

# Faceta A — Tecnología (IEEE/ACM)
ai = add_preferred_with_synonyms(
    builder,
    "artificial intelligence",
    synonyms=["AI"],
    scope_note="Computational methods that emulate or augment intelligent behavior",
)

ml = add_preferred_with_synonyms(builder, "machine learning")
ml.add_relation(RelationType.BT, ai)

dl = add_preferred_with_synonyms(builder, "deep learning")
dl.add_relation(RelationType.BT, ml)

cnn = add_preferred_with_synonyms(builder, "convolutional neural networks", synonyms=["CNN"])
cnn.add_relation(RelationType.BT, dl)
rnn = add_preferred_with_synonyms(builder, "recurrent neural networks", synonyms=["RNN"])
rnn.add_relation(RelationType.BT, dl)
transformers = add_preferred_with_synonyms(builder, "transformers", synonyms=["transformer model*"])
transformers.add_relation(RelationType.BT, dl)

cv = add_preferred_with_synonyms(builder, "computer vision")
cv.add_relation(RelationType.BT, ai)
img_proc = add_preferred_with_synonyms(builder, "image processing")
img_proc.add_relation(RelationType.BT, cv)

remote_sensing = add_preferred_with_synonyms(builder, "remote sensing")
remote_sensing.add_relation(RelationType.RT, cv)
hyperspectral = add_preferred_with_synonyms(builder, "hyperspectral imaging")
hyperspectral.add_relation(RelationType.BT, remote_sensing)
multispectral = add_preferred_with_synonyms(builder, "multispectral imaging")
multispectral.add_relation(RelationType.BT, remote_sensing)
nir = add_preferred_with_synonyms(builder, "near infrared spectroscopy", synonyms=["NIR"])
nir.add_relation(RelationType.RT, hyperspectral)

svm = add_preferred_with_synonyms(builder, "support vector machine", synonyms=["SVM"])
svm.add_relation(RelationType.RT, ml)
rf = add_preferred_with_synonyms(builder, "random forest")
rf.add_relation(RelationType.RT, ml)

# Faceta B — Enfermedad/estrés (AGROVOC/NAL/CAB)
plant_health = add_preferred_with_synonyms(builder, "plant health")
plant_diseases = add_preferred_with_synonyms(builder, "plant diseases", synonyms=["plant disease*", "crop disease*"])
plant_diseases.add_relation(RelationType.BT, plant_health)

pathogens = add_preferred_with_synonyms(builder, "plant pathogens", synonyms=["pathogen*"])
pathogens.add_relation(RelationType.RT, plant_diseases)

fungal = add_preferred_with_synonyms(builder, "fungal diseases")
fungal.add_relation(RelationType.BT, plant_diseases)
bacterial = add_preferred_with_synonyms(builder, "bacterial diseases")
bacterial.add_relation(RelationType.BT, plant_diseases)
viral = add_preferred_with_synonyms(builder, "viral diseases")
viral.add_relation(RelationType.BT, plant_diseases)

disease_severity = add_preferred_with_synonyms(builder, "disease severity")
disease_severity.add_relation(RelationType.RT, plant_diseases)
disease_incidence = add_preferred_with_synonyms(builder, "disease incidence")
disease_incidence.add_relation(RelationType.RT, plant_diseases)
disease_detection = add_preferred_with_synonyms(builder, "disease detection")
disease_detection.add_relation(RelationType.RT, plant_diseases)

# Faceta C — Predicción/modelado
prediction = add_preferred_with_synonyms(builder, "prediction", synonyms=["forecasting", "estimation"])
early_detection = add_preferred_with_synonyms(builder, "early detection", synonyms=["early warning"])
early_detection.add_relation(RelationType.BT, prediction)
monitoring = add_preferred_with_synonyms(builder, "monitoring", synonyms=["real-time monitoring"])
monitoring.add_relation(RelationType.RT, early_detection)

regression = add_preferred_with_synonyms(builder, "regression")
regression.add_relation(RelationType.RT, prediction)
classification = add_preferred_with_synonyms(builder, "classification")
classification.add_relation(RelationType.RT, prediction)

# Resultado auxiliar (rendimiento/calidad general) — se deja construido para poder activar la versión "nutrición + rendimiento"
crop_production = add_preferred_with_synonyms(builder, "crop production")
crop_yield = add_preferred_with_synonyms(builder, "crop yield", synonyms=["yield"])
crop_yield.add_relation(RelationType.BT, crop_production)
fruit_yield = add_preferred_with_synonyms(builder, "fruit yield")
fruit_yield.add_relation(RelationType.BT, crop_yield)
marketable_yield = add_preferred_with_synonyms(builder, "marketable yield")
marketable_yield.add_relation(RelationType.RT, crop_yield)

fruit_quality = add_preferred_with_synonyms(builder, "fruit quality", synonyms=["harvest quality", "quality assessment"])
fruit_quality.add_relation(RelationType.BT, crop_production)

# Calidad fisicoquímica/organoléptica (típica en calidad de fruto)
brix = add_preferred_with_synonyms(builder, "total soluble solids", synonyms=["Brix", "soluble solids"])
brix.add_relation(RelationType.RT, fruit_quality)
firmness = add_preferred_with_synonyms(builder, "firmness")
firmness.add_relation(RelationType.RT, fruit_quality)
color = add_preferred_with_synonyms(builder, "color")
color.add_relation(RelationType.RT, fruit_quality)
fruit_weight = add_preferred_with_synonyms(builder, "fruit weight")
fruit_weight.add_relation(RelationType.RT, fruit_quality)

# Faceta D (principal) — Matriz nutricional / calidad nutricional (macro + micro + minerales + fitonutrientes)
nutritional_quality = add_preferred_with_synonyms(
    builder,
    "nutritional quality",
    synonyms=[
        "nutritional value",
        "nutritional composition",
        "nutrient content",
        "nutritional profile",
        "nutrient profile",
        "food composition",
    ],
)

# Para no perder estudios que usen lenguaje de "fruit quality" sin decir nutrición, lo dejamos como término relacionado (no obligatorio).
fruit_quality.add_relation(RelationType.RT, nutritional_quality)

# Términos de nutrición frecuentes en Scopus (a veces aparecen sin decir "nutritional quality")
proximate_composition = add_preferred_with_synonyms(
    builder,
    "proximate composition",
    synonyms=["proximate analysis", "chemical composition"],
)
proximate_composition.add_relation(RelationType.BT, nutritional_quality)

nutritional_analysis = add_preferred_with_synonyms(
    builder,
    "nutritional analysis",
    synonyms=["nutrient analysis"],
)
nutritional_analysis.add_relation(RelationType.BT, nutritional_quality)

macronutrients = add_preferred_with_synonyms(builder, "macronutrients")
macronutrients.add_relation(RelationType.BT, nutritional_quality)
protein = add_preferred_with_synonyms(builder, "protein content", synonyms=["protein"])
protein.add_relation(RelationType.BT, macronutrients)
carbohydrate = add_preferred_with_synonyms(builder, "carbohydrate content", synonyms=["carbohydrate*", "sugar content"])
carbohydrate.add_relation(RelationType.BT, macronutrients)
lipid = add_preferred_with_synonyms(builder, "lipid content", synonyms=["fat content", "lipid*"])
lipid.add_relation(RelationType.BT, macronutrients)
fiber = add_preferred_with_synonyms(builder, "dietary fiber", synonyms=["fibre", "fiber content"])
fiber.add_relation(RelationType.BT, macronutrients)

micronutrients = add_preferred_with_synonyms(builder, "micronutrients")
micronutrients.add_relation(RelationType.BT, nutritional_quality)
vitamins = add_preferred_with_synonyms(builder, "vitamins", synonyms=["vitamin content"])
vitamins.add_relation(RelationType.BT, micronutrients)
vitamin_c = add_preferred_with_synonyms(builder, "vitamin C", synonyms=["ascorbic acid"])
vitamin_c.add_relation(RelationType.BT, vitamins)

minerals = add_preferred_with_synonyms(builder, "minerals", synonyms=["mineral content", "mineral composition"])
minerals.add_relation(RelationType.BT, micronutrients)
trace_elements = add_preferred_with_synonyms(builder, "trace elements", synonyms=["microminerals"])
trace_elements.add_relation(RelationType.BT, minerals)
iron = add_preferred_with_synonyms(builder, "iron", synonyms=["Fe"])
iron.add_relation(RelationType.BT, minerals)
zinc = add_preferred_with_synonyms(builder, "zinc", synonyms=["Zn"])
zinc.add_relation(RelationType.BT, minerals)
calcium = add_preferred_with_synonyms(builder, "calcium", synonyms=["Ca"])
calcium.add_relation(RelationType.BT, minerals)
potassium = add_preferred_with_synonyms(builder, "potassium", synonyms=["K"])
potassium.add_relation(RelationType.BT, minerals)

phytochemicals = add_preferred_with_synonyms(builder, "phytochemicals", synonyms=["phytonutrients", "bioactive compounds"])
phytochemicals.add_relation(RelationType.BT, nutritional_quality)
phenolic_compounds = add_preferred_with_synonyms(
    builder,
    "phenolic compounds",
    synonyms=["polyphenols", "phenolic content", "total phenolic content", "TPC"],
)
phenolic_compounds.add_relation(RelationType.BT, phytochemicals)
flavonoids = add_preferred_with_synonyms(
    builder,
    "flavonoids",
    synonyms=["flavonoid content", "total flavonoid content", "TFC"],
)
flavonoids.add_relation(RelationType.BT, phytochemicals)
anthocyanins = add_preferred_with_synonyms(builder, "anthocyanins", synonyms=["anthocyanin content"])
anthocyanins.add_relation(RelationType.BT, phytochemicals)
carotenoids = add_preferred_with_synonyms(builder, "carotenoids", synonyms=["carotenoid content"])
carotenoids.add_relation(RelationType.BT, phytochemicals)
lycopene = add_preferred_with_synonyms(builder, "lycopene")
lycopene.add_relation(RelationType.BT, carotenoids)
antioxidant_activity = add_preferred_with_synonyms(
    builder,
    "antioxidant activity",
    synonyms=["antioxidant capacity", "ORAC", "DPPH", "FRAP", "ABTS"],
)
antioxidant_activity.add_relation(RelationType.RT, phytochemicals)

# Faceta E — Dominio (ciclo corto / horticultura)
horticulture = add_preferred_with_synonyms(builder, "horticultural crops", synonyms=["horticulture", "horticultur*"])
vegetables = add_preferred_with_synonyms(builder, "vegetable crops", synonyms=["vegetable crop*"])
vegetables.add_relation(RelationType.BT, horticulture)
fruit_crops = add_preferred_with_synonyms(builder, "fruit crops", synonyms=["fruit crop*"])
fruit_crops.add_relation(RelationType.BT, horticulture)
annual_crops = add_preferred_with_synonyms(builder, "annual crops", synonyms=["short-season", "short season", "short-cycle", "short cycle"])
annual_crops.add_relation(RelationType.RT, horticulture)

# Faceta D alternativa (para más recall): resultados = (nutrición OR calidad)
nutrition_or_quality = add_preferred_with_synonyms(
    builder,
    "nutrition or quality outcomes",
    synonyms=["quality attributes", "quality parameters", "quality traits", "quality indicators"],
)
nutritional_quality.add_relation(RelationType.BT, nutrition_or_quality)
fruit_quality.add_relation(RelationType.BT, nutrition_or_quality)

def print_queries(title: str, facets, include_related: bool = False):
    builder.facets = list(facets)
    print(f"\n===== {title} =====")
    print("Ecuación (Scopus):")
    print(builder.build_search_equation(database="scopus", include_related=include_related))
    print("\nEcuación (WoS):")
    print(builder.build_search_equation(database="wos", include_related=include_related))
    print("\nEcuación (IEEE):")
    print(builder.build_search_equation(database="ieee", include_related=include_related))
    print("\nEcuación (ScienceDirect, simplificada):")
    print(builder.build_search_equation(database="sciencedirect", include_related=include_related))

# Versión 1 (recomendada): core nutrición (nutrición es obligatoria)
core_nutrition_facets = (ai, plant_health, prediction, nutritional_quality, horticulture)
print_queries("Core nutrición (nutrición-first)", core_nutrition_facets, include_related=False)

# Versión 2: core nutrición + términos relacionados (mayor recall; más ruido)
print_queries("Core nutrición + RT (más recall)", core_nutrition_facets, include_related=True)

# Versión 2b (Scopus-friendly): (nutrición OR calidad) + RT (más recall)
nutrition_or_quality_facets = (ai, plant_health, prediction, nutrition_or_quality, horticulture)
print_queries("Nutrición OR calidad + RT (más recall)", nutrition_or_quality_facets, include_related=True)

# Versión 3: nutrición + rendimiento (más estricta)
nutrition_plus_yield_facets = (ai, plant_health, prediction, nutritional_quality, crop_yield, horticulture)
print_queries("Nutrición + rendimiento (más estricta)", nutrition_plus_yield_facets, include_related=False)

print("\nEstructura por facetas (core nutrición):")
for facet in core_nutrition_facets:
    print(f"\nFaceta: {facet.term}")
    builder.print_thesaurus_structure(facet)
````
